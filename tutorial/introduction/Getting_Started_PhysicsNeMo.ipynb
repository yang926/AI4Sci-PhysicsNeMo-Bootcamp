## Introduction to PhysicsNeMo

PhysicsNeMo is an open-source PyTorch library for scientific and engineering deep-learning models and physics-based training. It supports PINNs, neural operators such as FNO and AFNO, graph and mesh models, data processing, and distributed training. This course pins its exercises to version 2.2.2.

[Official overview](https://docs.nvidia.com/physicsnemo/latest/index.html) · [Version 2.2.2 source](https://github.com/NVIDIA/physicsnemo/tree/v2.2.2) · [Version 2.0 migration guide](https://github.com/NVIDIA/physicsnemo/blob/v2.2.2/v2.0-MIGRATION-GUIDE.md)

### Building blocks of PhysicsNeMo

| Component | Role | Example in this course |
|---|---|---|
| `physicsnemo.models` | PyTorch model architectures | `FullyConnected`, followed by FNO and AFNO |
| `physicsnemo.sym` | Symbolic PDE definitions and residual evaluation | `PDE`, `PhysicsInformer` |
| Data and coordinate sampling | Prepare training inputs | `torch.rand`, the original Navier–Stokes array |
| Explicit training loop | Evaluate losses and optimize parameters | `torch.optim.Adam`, `backward`, `step` |
| Validation and inference | Check equations, boundary conditions, and prediction error | Fixed evaluation coordinates, `.npz`, Matplotlib |
| Additional components | Mesh, datapipes, distributed training, and more | Tools for scaling to larger problems |

**Sym stands for Symbolic.** The module uses SymPy expressions to define equations and evaluate residuals; not every PhysicsNeMo model requires Sym. These introductory MLP exercises also run on CPU. To use a CUDA GPU, install a compatible CUDA-enabled PyTorch environment and select `--device cuda`.

### Six-step approach to problem solving in PhysicsNeMo

1. **Problem and domain**: Specify input coordinates, physical quantities, parameter ranges, and units.
2. **Equations and model**: Define a SymPy `PDE` and a PyTorch `FullyConnected` network.
3. **Losses and sampling**: Make interior residual, initial condition, boundary condition, and data terms explicit.
4. **Evaluation coordinates**: Evaluate analytical solutions and residuals at fixed coordinates separate from training samples.
5. **Configuration**: Read the number of steps, batch size, learning rate, and network size from YAML.
6. **Training and inference**: Run the Adam loop and inspect the saved model, losses, and predictions.

The current course shows the training loop directly instead of hiding it inside framework-specific orchestration. Hydra remains an available general configuration tool, but these exercises do not require it.

## Physics-Informed approach vs Data-driven approach to neural networks

A **data-driven surrogate** minimizes the difference between predictions and simulation or observation targets. For example, an FNO can predict a pressure field from a permeability field. A **PINN** predicts physical quantities from coordinates and includes equation residuals and initial/boundary conditions in its loss. A **PINO** incorporates physics residuals into neural-operator training.

Distinguish model architecture from training method. `FullyConnected` and FNO are architectures; PINN and PINO describe ways to use physical information during training. Sym helps evaluate the residuals. For every approach, validate data and boundary-condition quality, equation assumptions, and generalization beyond the training domain separately.
