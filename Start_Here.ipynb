채점 연결 상태는 각 도전 과제의 마지막 제출 패널에서 확인하세요. 연결된 작업 공간에서는 닉네임을 등록하고 코드를 제출할 수 있습니다. 연결 없이도 로컬 실습과 저장된 코드 확인은 사용할 수 있습니다.

서울에서 열리는 AI for Science 학술대회의 실습형 튜토리얼입니다. 물리 정보 신경망(PINN)으로 미분방정식을 풀고, 사전 학습된 FourCastNet으로 기상을 예측하며, 장에서 장으로의 매핑을 학습하는 문제에서 FNO, AFNO, PINO를 비교합니다.

오전에는 PINN 학습 실습 세 개와 기상 추론 실습 하나를 통해 각 방법을 익힙니다. 오후에는 파동, 유체 흐름, 온도 수송, 신경 연산자를 다루는 개인 도전 과제 네 개를 풉니다. 이 자료는 PhysicsNeMo 2.2.2를 사용하며 Open Hackathons AI-Powered Physics Bootcamp를 바탕으로 구성했습니다. 실습 4는 강사의 승인을 받아 기상 예측 실습으로 대체했습니다.

권장 사전 지식은 기본적인 Python 함수와 배열, 1차·2차 미분, 초기 조건과 경계 조건입니다. PhysicsNeMo 사용 경험은 필요하지 않습니다.

[학술대회 웹사이트](https://ai4scikorea.org/)

실습에서는 완성된 프로그램을 제공합니다. 실습 1–3은 PINN을 학습하고, 실습 4는 옵티마이저 없이 사전 학습 모델을 실행합니다. 실습 4에서는 48시간 기상 예측을 생성하고 ERA5 재분석 자료와 비교하며, 예측 선행 시간이 늘어날 때 오차가 어떻게 달라지는지 살펴봅니다. 위 일정은 공지된 세션 제목의 의미와 시각을 유지했습니다.

도전 과제에서는 연결된 Python 파일을 수정하여 각 단계를 해결한 뒤, JupyterLab에서 실행하고 결과를 평가합니다.

오후 도전 과제는 개인별로 진행합니다. 현재 시범 채점에서는 학습 손실의 크기가 아니라 지정된 방정식, 조건, 형상, 모델을 구현했는지에 따라 점수를 부여합니다. 수치 오차는 별도의 피드백으로 제공합니다. 완료한 단계를 모두 함께 제출하세요. 빠진 단계는 0점이며, 서로 다른 제출의 점수를 합산하지 않습니다. 행사 규칙은 주최 측이 확정합니다. [평가 및 제출 안내](ETC/course_materials/ASSESSMENT.md)를 참고하세요.

## JupyterLab 사용 방법

1. 위 표에서 수업 노트북을 엽니다. 환경 설정 셀을 먼저 실행한 뒤 각 절을 순서대로 진행하세요. 코드 셀은 Shift+Enter로 실행합니다. 왼쪽 사이드바의 목차에서 각 절을 확인할 수 있습니다.
2. 도전 과제는 저장한 연습 함수만 실행합니다. 연결된 `.py` 파일을 노트북 옆에 열고, 현재 단계의 함수를 완성한 뒤 저장하세요. 연습을 끝내기 전에는 Run All(모두 실행)을 사용하지 마세요. 도전 과제 4는 학습 전에 데이터 생성 단계도 실행해야 합니다.
3. 학습 전에도 마지막 제출 패널을 열 수 있습니다. 해당 코드 셀을 실행하고 완료한 단계를 선택한 뒤 **Check saved code(저장된 코드 확인)**를 클릭하세요. 학습하거나 채점기에 보내지 않고 구현을 확인합니다. Ctrl/Cmd 키를 누른 채 클릭하면 미완료 단계의 선택을 해제할 수 있습니다. 확인 후 현재 단계의 실행 셀과 결과 셀로 돌아가세요.
4. 학습 실습에서는 학습 전후의 오차와 그래프를 비교하세요. 자세한 내용은 Full metrics and run settings(전체 지표 및 실행 설정)를 엽니다. 실습 4에서는 예측, ERA5 재분석, 지속성 예측의 오차를 비교합니다. 실행이 성공했다는 사실만으로 정확성을 입증할 수는 없습니다.
5. 도전 과제를 제출하려면 마지막 패널의 **Nickname(닉네임)**에 이름을 입력하고 **Register nickname(닉네임 등록)**을 한 번 클릭하세요. 닉네임은 네 도전 과제에서 공통으로 사용합니다. 완료한 단계를 모두 선택하고 **Submit code(코드 제출)**를 클릭하세요. 노트북 셀을 실행하는 것만으로는 제출되지 않습니다.

Markdown 파일은 Open With → Markdown Preview(다음으로 열기 → Markdown 미리 보기)로 읽을 수 있습니다. 셀 실행이 실패하면 첫 번째 오류를 해결하고 셀을 다시 실행한 뒤 그래프를 그리세요. 이전 그래프는 실패한 실행의 결과가 아닙니다.

## PhysicsNeMo 코드 따라가기

PINN 수업에서는 빈칸에 들어갈 방정식뿐 아니라 모델 예측이 물리 손실과 매개변수 갱신으로 이어지는 과정을 익힙니다. 실습 4에서는 사전 학습된 AFNO 모델이 현재 기상 상태에서 다음 상태를 예측하는 과정을 살펴봅니다. 각 수업은 PhysicsNeMo가 제공하는 기능, PyTorch가 계산하는 내용, 강의 보조 함수가 연결하는 부분을 구분해 설명합니다.

도전 과제 1–3의 `create_model`, `create_informer`, `residuals`는 PhysicsNeMo의 공개 API가 아니라 강의 보조 함수입니다. [코드 해설](ETC/course_materials/PHYSICSNEMO_WORKFLOW.md)을 따라 `FullyConnected`와 `PhysicsInformer`까지 살펴보세요. 각 실습과 도전 과제 4에서는 해당 수업의 모델과 잔차 계산 경로를 따로 설명합니다.

도전 과제를 실행하기 전에 모델의 입력과 출력을 확인하고, 방정식 및 조건 손실의 위치와 PyTorch 옵티마이저의 갱신 단계를 찾아보세요. 이러한 코드 읽기 점검 항목은 연습 내용이나 채점 규칙을 바꾸지 않습니다.
