A hands-on tutorial at the AI for Science conference in Seoul. Use PhysicsNeMo to solve differential equations with physics-informed neural networks (PINNs), then compare FNO, AFNO and PINO on a field-to-field learning problem.

The morning introduces the methods through four worked Labs. In the afternoon, solve four individual Challenges covering waves, fluid flow, temperature transport and neural operators. The materials use PhysicsNeMo 2.2.2 and are adapted from the Open Hackathons AI-Powered Physics Bootcamp.

[Conference website](https://ai4scikorea.org/)

## Working in JupyterLab

1. Open a lesson from the table above. Use Shift+Enter to run a code cell. The left sidebar's table of contents lists the sections.
2. Keep the notebook and its linked `.py` file side by side. Save the Python file before rerunning a Challenge.
3. Compare the before/after errors and plot. Open Full metrics and run settings for details. A short run can confirm that the code works without producing an accurate solution.

To read a Markdown file, select Open With → Markdown Preview. If a cell fails, fix its first error and rerun it before plotting. An old plot does not show the result of a failed run.

## Folder guide

`01_labs/` and `02_challenges/` follow the course order. Each numbered folder contains one lesson notebook, its programs and supporting files. `ETC/` contains shared runtime code, environment setup and instructor tools. Keep it in place: the lessons import its code.
