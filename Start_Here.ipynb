A hands-on tutorial at the AI for Science conference in Seoul. Solve differential equations with physics-informed neural networks (PINNs), generate weather forecasts with pretrained FourCastNet, and compare FNO, AFNO and PINO on a field-to-field learning problem.

The morning introduces the methods through three PINN training Labs and a weather inference Lab. In the afternoon, solve four individual Challenges covering waves, fluid flow, temperature transport and neural operators. The materials use PhysicsNeMo 2.2.2 and are adapted from the Open Hackathons AI-Powered Physics Bootcamp; Lab 4 is an instructor-approved weather forecasting replacement.

Recommended background: basic Python functions and arrays, first and second derivatives, and initial/boundary conditions. No prior PhysicsNeMo experience is assumed.

[Conference website](https://ai4scikorea.org/)

The Labs provide complete programs. Labs 1–3 train PINNs; Lab 4 runs a pretrained model without an optimizer. In Lab 4, generate a 48-hour weather forecast, compare it with ERA5 reanalysis, and inspect errors as lead time increases. The published session title and times above are retained.

In the Challenges, edit the linked Python files to solve each level, then run and evaluate your solution in JupyterLab.

The afternoon Challenges are individual exercises. The current pilot awards points for implementing the assigned equations, conditions, geometry and models, not for the lowest training loss. Numerical errors are separate feedback. Submit all completed Levels together; omitted Levels score zero and separate attempts are not combined. The organizers confirm the event rules. See [evaluation and submission](ETC/course_materials/ASSESSMENT.md).

## Working in JupyterLab

1. Open a lesson from the table above. Run its setup cell first, then follow the sections in order. Use Shift+Enter to run a code cell. The left sidebar's table of contents lists the sections.
2. Challenges run only your saved exercise functions. Open the linked `.py` file beside the notebook, complete the current Level's functions, and save the file. Do not use Run All while exercises are unfinished. Challenge 4 also requires its data-generation step before training.
3. You can open the final submission panel before training: run its code cell, select the completed Levels, and click **Check saved code**. This checks the implementation without training or sending it to the judge. Ctrl/Cmd-click to deselect unfinished Levels. Return to the current Level's execution and result cells.
4. For training exercises, compare the before/after errors and plot; open Full metrics and run settings for details. Lab 4 instead compares forecast, ERA5 reanalysis and persistence errors. A successful run alone does not establish accuracy.
5. To submit a Challenge, enter your **Nickname** in the final panel and click **Register nickname** once. The name is shared across all four Challenges. Select all completed Levels and click **Submit code**; running notebook cells alone does not submit anything.

To read a Markdown file, select Open With → Markdown Preview. If a cell fails, fix its first error and rerun it before plotting. An old plot does not show the result of a failed run.

## Follow the PhysicsNeMo code

In the PINN lessons, learn how a model prediction becomes a physics loss and a parameter update, not only which equation fills a blank. Lab 4 instead follows a pretrained AFNO model from the current weather state to the next. Each lesson identifies what PhysicsNeMo supplies, what PyTorch computes, and what the course helpers connect.

In Challenges 1–3, `create_model`, `create_informer` and `residuals` are course helpers, not public PhysicsNeMo API names. Follow them into `FullyConnected` and `PhysicsInformer` with the [code walkthrough](ETC/course_materials/PHYSICSNEMO_WORKFLOW.md). The Labs and Challenge 4 show their own model and residual paths.

Before running a Challenge, name the model inputs and outputs, locate the equation and condition losses, and find the PyTorch optimizer step. These reading checkpoints do not change the exercise or scoring rules.

## Folder guide

`01_labs/` and `02_challenges/` follow the course order. Each numbered folder contains one lesson notebook, its programs and supporting files. `ETC/` contains shared runtime code, environment setup and instructor tools. Keep it in place: the lessons import its code.
