# AI4Sci PhysicsNeMo Bootcamp

Complete the Labs and Challenges with PhysicsNeMo 2.2.2.

1. Run the [environment check](ai4sci/00_environment_check.ipynb).
2. Read the [course guide: edit, save, run and evaluate](ai4sci/README.md).
3. Follow the lessons below.

For setup, see the [deployment guide](Deployment_Guide.MD).

## Complete course

1. [Introduction to PhysicsNeMo](tutorial/introduction/Getting_Started_PhysicsNeMo.ipynb)
2. [PINN fundamentals](tutorial/introduction/Introductory_Notebook.ipynb)
3. [Projectile ODEs](tutorial/projectile/Getting_Started_Projectile.ipynb)
4. [Diffusion PDEs](tutorial/diffusion_1d/Diffusion_Problem_Notebook.ipynb)
5. [Navier–Stokes](tutorial/navier_stokes/Weather-forecasting-navier-stokes.ipynb)
6. [Wave dynamics](challenge/wave/Advanced_Wave_Dynamics.ipynb)
7. [Fluid flow](challenge/fuild/Fluid_Structure_Interaction.ipynb)
8. [Climate modeling](challenge/climate/Multi-Physics_Climate_Modeling.ipynb)
9. [Neural operators](challenge/neural_operator/Advanced_Neural_Operators.ipynb)

For each Challenge, complete the `student_*` function in the actual `.py` file, save it, then run the notebook execution cell. `--reference` selects the instructor implementation. FNO, AFNO and PINO are three levels of the Neural Operators Challenge.
